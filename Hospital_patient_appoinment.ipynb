```python
import pandas as pd
```

```python
df = pd.read_csv("./Data/appointsments_data.csv")
df.head()
```

```python
df.info()
```

```python
#Convert dateitme information
df["appointment_date"] = pd.to_datetime(
    df["appointment_date"],
    format = "%d-%m-%Y %H:%M"
)
df.info()
```

```python
df["Date"] = df["appointment_date"].dt.date
```

```python
df["Date"] = df["appointment_date"].dt.date
df["Hour"] = df["appointment_date"].dt.hour
df["Weekday"] = df["appointment_date"].dt.day_name()
df["Month"] = df["appointment_date"].dt.month_name()
df.head()
```

```python
df.iloc[0]
```

```python
df = df.set_index("appointment_date")
df.head()
```

```python
df = df.sort_index()
df.head()
```

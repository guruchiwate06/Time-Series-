```python
from datetime import datetime 
now = datetime.now()
print(now)
```

```python
print("year:",now.year)
print("Month:",now.month)
print("Day:",now.day)
print("hour:",now.hour)
print("minute:",now.minute)
print("second:",now.second)
```

```python
print(now.strftime("%d-%m-%Y"))#date
print(now.strftime("%d/%m/%Y"))#date
print(now.strftime("%d-%m-%Y %H:%M:%S"))#Date and time 
print(now.strftime("%d-%B-%Y"))#Month in char
print(now.strftime("%A"))#Day
print(now.strftime("%j-%B-%Y"))#Exp
print(now.strftime("%A-%b-%y"))#Exp
print(now.strftime("%d-%B-%y"))#Exp
```

```python
from datetime import timedelta
today = date.today()
after10days = today + timedelta(days=10)
before10days = today - timedelta(days=10)
print("date after 10 days:",after10days)
print("date before 10 days:",before10days)
```

```python
from datetime import datetime, time, date

start = time(9,30)
end  = time(17,30)
start_dt = datetime.combine(date.today(),start)
end_dt = datetime.combine(date.today(),end)

difference = end_dt - start_dt

print("Difference:" , difference)
print("Hours:",difference.total_seconds()/3600)
```

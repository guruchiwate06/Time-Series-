```python
import pandas as pd 
import matplotlib.pyplot as plt
from   statsmodels.tsa.seasonal import seasonal_decompose
```

```python
df = pd.read_csv("tourist_arrivals_5_years.csv")
df.head()
```

```python
df['Date'] = pd.to_datetime(df['Date'])
df.info()
```

```python
df = df.set_index("Date")
```

```python
df = df.sort_index()
df
```

```python
plt.figure(figsize=(12, 4))
plt.plot(df.index, df['Tourist_Arrivals'], label='Tourist Arrivals')
plt.xlabel('Date')
plt.ylabel('Tourist Arrivals')
plt.title('Tourist Arrivals Over Time')
plt.legend()
plt.show()
```

```python
result =seasonal_decompose(
    df["Tourist_Arrivals"],
    model = "multiplicative",
    period =12
)
```

```python
result.observed
```

```python
result.trend
```

```python
result.seasonal
```

```python
result.resid
```

```python
#Create a dataframe from decomposition components
decomp_df = pd.DataFrame({
    "Observed" : result.observed,
    "Trend" : result.trend,
    "Seasonal" : result.seasonal,
    "Residual" : result.resid
})

decomp_df.head()
```

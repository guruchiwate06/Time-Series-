```python
import pandas as pd 
import matplotlib.pyplot as plt 

df = pd.read_csv("electricity_data1.csv")
df.head()
```

```python
df.info()
```

```python
df["Date"] = pd.to_datetime(df["Date"],dayfirst=True)#the date is not in proper format so we use dayfirst
df.set_index("Date")
df.sort_index(inplace=True)
df.info()
```

```python
df.head()
```

```python
plt.figure(figsize=(12,5))

plt.plot(
    df["Date"],
    df["Electricity_MWh_1"]
)

plt.title("Original Electricity Consumption")
plt.xlabel("Date")
plt.ylabel("Electricity Consumption (MWh)")
plt.grid(True)
plt.show()
```

```python
#Check stationary 
stationary_monthly_stats = (
    #Group the data month by month using Data column
    df.resample("ME", on="Date")["Electricity_MWh_1"].agg(["mean","std"])
)
print("\nStationary Data")
stationary_monthly_stats.head(10)
```

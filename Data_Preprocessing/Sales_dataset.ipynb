```python
import pandas as pd
import numpy as np


data = {
    "Date": pd.date_range(start="2024-01-01", periods=10, freq="D"),
    "Sales": [200, 220, np.nan, 250, 210, np.nan, 300, 280, np.nan, 260]
}

df = pd.DataFrame(data)
print(df)
df.to_csv("toy_dataset")
```

#### Forward Fill

```python
fwd_fill = df["Sales"].ffill()
```

```python
print(fwd_fill)
```

### Backward fill

```python
b_fill = df["Sales"].bfill()
```

```python
print(b_fill)
```

```python
inter = df["Sales"].interpolate()
```

```python
print(inter)
```

```python
import pandas as pd
import numpy as np


data = {
    "Date": pd.date_range(start="2024-01-01", periods=10, freq="D"),
    "Sales": [np.nan, 220, np.nan, 250, 210, np.nan, 300, 280,260,np.nan]
}

df = pd.DataFrame(data)
print(df)
df.to_csv("toy_dataset")
```

```python
#IF the null value is present at the end of the data 

end_inter = df["Sales"].interpolate(limit_direction="backward")
print(end_inter)
```

```python
top_inter = df["Sales"].interpolate(limit_direction="forward")
print(top_inter)
```

```python
both_inter = df["Sales"].interpolate(limit_direction="both")
print(both_inter)
```

```python
end_inter = df["Sales"].interpolate(limit_direction="backward")
print(end_inter)
```
